<h1 style=" color:gold; font-size:250%; font-weight:bold"> 1)Common Method using open() function </h1>

<h style="color:coral; font-size:150%; font-weight:bold">1.A) Checking Important attributes </h> 

```python
f=open('example.txt', 'r')
#Important attributes to check the files 
print(f.name)
print(f.mode)
print(f.encoding)
print(f.closed)
f.close() 
```

<h style="color:coral; font-size:150%; font-weight:bold">1.B) Peforming actions: READING FILE </h> 

```python
#Reading file in different 3-scenarios

#a) Reading the whole file as a single string 
f=open('example.txt', 'r')
f_contents =f.read() # Action: reading the file and storing contents in a variable 
# print(f_contents)
f.close()

#b)f.readline():: Reading line by line 
f=open('example.txt', 'r')
line1=f.readline() 
print(line1, end='') # Reads and prints the 1st line
line2=f.readline() 
print(line2) # Reads and prints the 2nd line

#b)f.readlines():: Readding all the lines and return as list[] of strings 
f=open('example.txt', 'r')
f_content_lines=f.readlines()
for line in f_content_lines:
    print(line, end='') #pass empty string to avoid space b/w two line 
```

<h1 style=" color:gold; font-size:250%; font-weight:bold"> 2)Context Manager Method </h1>

<h style="color:coral; font-size:150%; font-weight:bold">1.A)READING FILE </h> 

```python
#A) Method-1: f.read()
with open('example.txt', 'r') as f:
    f_contents=f.read()
    print(f_contents)

#B) Method-2: using f.readline()
with open('example.txt', 'r') as f:
    f_contents=f.readline()
    print(f_contents)
    f_contents=f.readline()
    print(f_contents)

#C) Method-3: using f.readlines()
with open('example.txt', 'r') as f:
    f_contents=f.readlines()
    print(f_contents)
```

<h style="color:coral; font-size:150%; font-weight:bold">Using loop for READING FILE </h> 

```python
with open('example.txt', 'r') as f:
    size_to_read=10 #instead of doing of whole at once do in chunk size of 10 
    f_contents=f.read(size_to_read) #sise to be read at a time
    while len(f_contents)>0:  #while size of the content read >o then
        print(f_contents, end='')
        f_contents=f.read(size_to_read) #mentioning the next chunk size to be read
```

<h style="color:coral; font-size:150%; font-weight:bold">1.A)WRITING FILE </h> 

```python
#A)only wrting the files
with open('example_only_write.txt', 'w') as f:  #mode w:with overwite(if file exist)else create &write
    f.write('Writing Test')

#B) Read and write mode 
with open ('example.txt', 'r') as rf:
    with open('example_read_write.txt', 'w') as wf:
        for line in rf:
            wf.write(line)
```
